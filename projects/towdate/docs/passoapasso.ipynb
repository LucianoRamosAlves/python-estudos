Uma analogia para o ToDate
Imagine que o ToDate é um restaurante:
🏢 Flask → o prédio.
⚙️ extensions.py → a sala onde ficam os equipamentos (fogão, geladeira, forno).
🧠 init.py → o gerente que liga tudo e organiza o restaurante antes de abrir.
🗄️ SQLAlchemy → o estoque (onde os dados ficam organizados).
🔄 Flask-Migrate → a equipe de reforma, que altera o restaurante sem demolir tudo.
🔑 Flask-Login → o porteiro, que sabe quem pode entrar na área VIP (páginas privadas).
